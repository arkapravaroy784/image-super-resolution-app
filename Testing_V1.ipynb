```python
import os
from PIL import Image
from torch.utils.data import Dataset

class DIV2KDataset(Dataset):
    def __init__(self, hr_dir, lr_dir, transform=None):
        self.hr_dir = hr_dir
        self.lr_dir = lr_dir
        self.files = sorted(os.listdir(hr_dir))
        self.transform = transform

    def __len__(self):
        return len(self.files)

    def __getitem__(self, idx):
        hr = Image.open(os.path.join(self.hr_dir, self.files[idx])).convert("RGB")
        lr = Image.open(os.path.join(self.lr_dir, self.files[idx])).convert("RGB")

        if self.transform:
            hr = self.transform(hr)
            lr = self.transform(lr)

        return lr, hr
```

```python
import random
import torchvision.transforms.functional as F
from torch.utils.data import Dataset
from PIL import Image

class DIV2KPatchDataset(Dataset):
    def __init__(self, hr_dir, lr_dir, patch_size=48, scale=4):
        self.hr_dir = hr_dir
        self.lr_dir = lr_dir
        self.files = sorted(os.listdir(hr_dir))
        self.patch_size = patch_size
        self.scale = scale

    def __len__(self):
        return len(self.files)

    def __getitem__(self, idx):
        hr_name = self.files[idx]                 # e.g. 0196.png
        lr_name = hr_name.replace(".png", "x4.png")

        hr = Image.open(os.path.join(self.hr_dir, hr_name)).convert("RGB")
        lr = Image.open(os.path.join(self.lr_dir, lr_name)).convert("RGB")


        lr_w, lr_h = lr.size
        x = random.randint(0, lr_w - self.patch_size)
        y = random.randint(0, lr_h - self.patch_size)

        lr_patch = F.crop(lr, y, x, self.patch_size, self.patch_size)
        hr_patch = F.crop(
            hr,
            y * self.scale,
            x * self.scale,
            self.patch_size * self.scale,
            self.patch_size * self.scale
        )

        lr_patch = T.ToTensor()(lr_patch)
        hr_patch = T.ToTensor()(hr_patch)

        return lr_patch, hr_patch
```

```python
from torch.utils.data import DataLoader

patch_dataset = DIV2KPatchDataset(
    hr_dir="/content/DIV2K_train_HR",
    lr_dir="/content/DIV2K_train_LR_bicubic/X4",
    patch_size=48,
    scale=4
)

patch_loader = DataLoader(
    patch_dataset,
    batch_size=16,
    shuffle=True,
    num_workers=2,
    pin_memory=True
)
```

```python
model.train()

for epoch in range(2):  # start small
    total_loss = 0

    for lr, hr in patch_loader:
        lr = lr.to(device, non_blocking=True)
        hr = hr.to(device, non_blocking=True)

        optimizer.zero_grad()
        sr = model(lr)
        loss = criterion(sr, hr)
        loss.backward()
        optimizer.step()

        total_loss += loss.item()

    print(f"Epoch {epoch+1} | Loss: {total_loss / len(patch_loader):.4f}")
```
